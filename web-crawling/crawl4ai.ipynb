```python
import asyncio
from crawl4ai import AsyncWebCrawler

async def fetch(url: str):
    async with AsyncWebCrawler() as crawler:
        result = await crawler.arun(
            url=url,
        )
        return result.markdown
```

```python
content = await fetch("https://www.nbcnews.com/business")
content
```

```python
content = await fetch(url="https://www.canada.ca/en/immigration-refugees-citizenship/services/immigrate-canada/express-entry.html")
content
```
